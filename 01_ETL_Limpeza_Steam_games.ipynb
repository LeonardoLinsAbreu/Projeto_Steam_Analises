# 7. Preenchimento de Valores Nulos e Remoção de Registros Inválidos

Nesta etapa, criei uma lista chamada **"colunas_texto"**, para receber os valores das colunas de texto, em seguida, . Novamente criei outra lista chamada **"colunas_numericas"**, para receber os valores das colunas numéricas. Em seguida, algumas informações sobre outros códigos usados nesta linha:


1.   **Preenchimento de Nulos em Texto (.fillna('Desconhecido')):** Substitui valores nulos nas colunas descritivas (como Desenvolvedores, Gêneros e Links) pela flag 'Desconhecido'. Isso evita a exclusão de linhas inteiras apenas por falta de informações secundárias.
2.   **Preenchimento de Nulos Numéricos (.fillna(0)):** Atribui o valor 0 aos registros nulos nas colunas numéricas de identificação e ranqueamento (AppID e Score rank), garantindo que não ocorram erros de cálculo ao importar os dados para o Power BI.
3.   **Remoção de Registros Inválidos (.dropna(subset=['...'])):** Exclui completamente qualquer linha que não possua o nome do jogo preenchido, pois um registro sem título, não possui validade analítica para o dashboard no Power BI.


```python
# Colunas de Texto (Developers, Publishers, Genres...)
# Em vez de apagar, preenchemos com 'Desconhecido' os valores nulos (em branco)

colunas_texto = ['User score','Windows', 'Supported languages','Developers', 'Publishers', 'Genres', 'Tags',
                 'Categories', 'Screenshots', 'Movies', 'Notes', 'About the game', 'Reviews', 'Metacritic url',
                 'Support email', 'Support url', 'Website', 'Header image', 'Average playtime forever'
                 ]

for col in colunas_texto:
    if col in df.columns:
        df[col] = df[col].fillna('Desconhecido')

# Colunas Numéricas (AppID e Score rank)
# Preenchemos com 0 os valores nulos (em branco) para não quebrar médias/somas no Power BI

colunas_numericas = ['Achievements', 'AppID', 'Score rank']

for col in colunas_numericas:
    if col in df.columns:
        df[col] = df[col].fillna(0)

# Apagamos APENAS se o nome do jogo estiver em branco

if 'Name' in df.columns:
    df = df.dropna(subset=['Name'])
```

# 8. Nova Visão dos Valores Nulos

# 9. Limpeza e Formatação da Coluna de Preços

Nesta etapa, converti a coluna de preços (que estava sendo lida como texto/objeto) em valores numéricos reais, permitindo cálculos e agregações futuras no Power BI (como ticket médio ou receita). Em seguida, algumas informações sobre o comando utilizado, realizando três ações:



1.   **Limpeza de Caracteres Especiais (str.replace):** Substitui a string "Free" pelo número "0" e remove possíveis símbolos de moeda (como $), limpando a formatação visual.
2.   **Conversão de Tipo (pd.to_numeric):** Transforma os dados em formato de texto para números matematicamente calculáveis (float/int).
3.   **Tratamento de Erros (errors='coerce' e fillna(0)):** Força qualquer outro texto não mapeado a se tornar nulo (NaN) e, em seguida, preenche esses nulos com 0, garantindo a integridade matemática da coluna.



```python
# Garantir que o Preço seja numérico (se houver texto como 'Free', vira 0)

if 'Price' in df.columns:
    df['Price'] = pd.to_numeric(df['Price'].astype(str).str.replace('Free', '0').str.replace('$', ''), errors='coerce').fillna(0)
```

# 10. Ajuste e Criação de Novas Métricas e Categorização de Preço

Nesta etapa, realizei a conversão da coluna **'Release date'** para o formato de data (datetime), pois anteriormente, estava em formato de texto (Object). Em seguida, fiz a criação de novas colunas baseadas nos dados existentes chamadas **"Total ratings e Approval rate"** para saber o total de avaliações (realizando a soma de avaliações Positivas e Negativas) e taxa de aprovação em %. Sendo importantepara facilitar a construção de indicadores (KPIs) e filtros no Power BI, reduzindo a necessidade de cálculos complexos na ferramenta de visualização. Em seguida, algumas informações sobre o comando utilizado:


1.   **Volume de Engajamento (Total ratings):** Soma simples das avaliações positivas e negativas, criando uma métrica única que reflete a popularidade e o volume de interações do jogo.
2.   **Métrica de Aceitação (Approval rate):** Calcula o percentual de aprovação do jogo. Foi utilizado o **.fillna(0)** de forma estratégica para evitar erros matemáticos (divisão por zero) em jogos que não possuem nenhuma avaliação registrada, finalizando com **.round(1)** para manter o número limpo (com uma casa decimal).
3.   **Segmentação Categórica (Price category):** Criação de uma função customizada no Python **(categorizar_preco)** mapeada na coluna de preço através do método **.apply()**. Isso transforma um dado contínuo (preço exato) em faixas categóricas (Grátis, Barato, Médio, Caro), o que é ideal para criar filtros de segmentação (Slicers) no dashboard do Power BI.



```python
# Converter a coluna 'Release date' para o formato de data (datetime)

df['Release date'] = pd.to_datetime(df['Release date'], errors='coerce')

# Total de avaliações (Positivos + Negativos) no DataFrame

if 'Positive' in df.columns and 'Negative' in df.columns:
    df['Total ratings'] = df['Positive'] + df['Negative']

    # Taxa de Aprovação (%)
    # Evitamos divisão por zero se o jogo não tiver avaliações

    df['Approval rate'] = (df['Positive'] / df['Total ratings'] * 100).fillna(0).round(1)

# Nova categoria de Preço (Grátis, Barato, Médio, Caro)

def categorizar_preco(preco):
    if preco == 0:
        return 'Grátis'
    elif preco < 20:
        return 'Até R$ 20 / $20'
    elif preco < 60:
        return 'R$ 20 a 60'
    else:
        return 'Acima de R$ 60'

if 'Price' in df.columns:
    df['Price category'] = df['Price'].apply(categorizar_preco)
```

# 11. Limpeza e Formatação de Colunas

Nesta etapa, realizei a remoção de colunas que não agregam valor analítico ou visual ao escopo deste projeto (como URLs, links de imagens e textos longos descritivos). O objetivo dessa exclusão é otimizar o DataFrame, reduzindo o uso de memória e deixando o arquivo mais leve, focando exclusivamente nas variáveis que serão utilizadas na construção do dashboard no Power BI.

```python
# Apagaremos as colunas que não agregam visualmente no DataFrame

colunas_para_remover = ['Average playtime forever', 'Average playtime two weeks', 'Median playtime two weeks',
                        'About the game', 'Screenshots', 'Reviews', 'Header image', 'Website', 'Support url',
                        'Support email', 'Metacritic url', 'Movies', 'Supported languages', 'User score', 'Windows', 'Mac', 'Linux', 'Metacritic score']

df = df.drop(columns=[col for col in colunas_para_remover if col in df.columns])
```

# 12. Segunda Camada de Limpeza de Duplicadas

Nesta etapa, foi aplicada uma segunda camada de limpeza de duplicatas. Diferente da limpeza geral inicial (que remove apenas linhas 100% idênticas), este comando utiliza o parâmetro subset para focar exclusivamente na coluna **"Name"**.

Isso garante que cada título de jogo apareça apenas uma vez na base de dados. É
um tratamento preventivo fundamental para evitar que um mesmo jogo (lançado com IDs diferentes ou variações regionais) infle a contagem total de jogos ou distorça as médias de avaliações e preços no Power BI.

```python
# Checamos as linhas duplicadas por Nome no DataFrame

if 'Name' in df.columns:
    df = df.drop_duplicates(subset=['Name'])
```
